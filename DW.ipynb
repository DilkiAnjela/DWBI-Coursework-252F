```python
import pandas as pd
import os

# --------------------------------------------------
# 1. Load Dataset
# --------------------------------------------------

file_path = r"C:\Users\dilki\Downloads\DW\2020_al_data_kaggle_upload_new_old_syllabi.csv"

df = pd.read_csv(file_path)

print("\n========== DATASET PROFILE ==========\n")

# --------------------------------------------------
# 2. Basic Dataset Information
# --------------------------------------------------

print("File Name:", os.path.basename(file_path))
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())


# --------------------------------------------------
# 3. Data Types
# --------------------------------------------------

print("\n========== DATA TYPES ==========\n")

print(df.dtypes)


# --------------------------------------------------
# 4. First Records
# --------------------------------------------------

print("\n========== SAMPLE DATA ==========\n")

print(df.head())


# --------------------------------------------------
# 5. Missing Values
# --------------------------------------------------

print("\n========== MISSING VALUES ==========\n")

missing_values = df.isnull().sum()

missing_percentage = (
    df.isnull().sum() / len(df) * 100
).round(2)

missing_report = pd.DataFrame({
    "Missing_Count": missing_values,
    "Missing_Percentage": missing_percentage
})

print(missing_report)


# --------------------------------------------------
# 6. Duplicate Rows
# --------------------------------------------------

print("\n========== DUPLICATES ==========\n")

duplicate_count = df.duplicated().sum()

print("Duplicate Rows:", duplicate_count)


# --------------------------------------------------
# 7. Unique Values
# --------------------------------------------------

print("\n========== UNIQUE VALUES ==========\n")

for column in df.columns:

    print(
        column,
        ":",
        df[column].nunique(dropna=False),
        "unique values"
    )


# --------------------------------------------------
# 8. Numeric Column Statistics
# --------------------------------------------------

print("\n========== NUMERIC STATISTICS ==========\n")

numeric_columns = df.select_dtypes(
    include=["int64", "float64"]
)

if len(numeric_columns.columns) > 0:

    print(numeric_columns.describe())

else:

    print("No numeric columns detected.")


# --------------------------------------------------
# 9. Categorical Column Statistics
# --------------------------------------------------

print("\n========== CATEGORICAL STATISTICS ==========\n")

categorical_columns = df.select_dtypes(
    include=["object"]
)

for column in categorical_columns.columns:

    print("\nColumn:", column)

    print(
        df[column]
        .value_counts(dropna=False)
        .head(10)
    )


# --------------------------------------------------
# 10. Check Blank Strings
# --------------------------------------------------

print("\n========== BLANK VALUES ==========\n")

for column in df.select_dtypes(include=["object"]).columns:

    blank_count = (
        df[column]
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    if blank_count > 0:

        print(column, ":", blank_count)


# --------------------------------------------------
# 11. Dataset Memory Usage
# --------------------------------------------------

print("\n========== MEMORY USAGE ==========\n")

memory_mb = (
    df.memory_usage(deep=True).sum()
    / 1024 ** 2
)

print(
    "Memory Usage:",
    round(memory_mb, 2),
    "MB"
)


# --------------------------------------------------
# 12. Generate Column Profile
# --------------------------------------------------

profile = []

for column in df.columns:

    profile.append({

        "column_name": column,

        "data_type":
            str(df[column].dtype),

        "total_rows":
            len(df),

        "non_null_count":
            df[column].notnull().sum(),

        "null_count":
            df[column].isnull().sum(),

        "null_percentage":
            round(
                df[column].isnull().mean() * 100,
                2
            ),

        "unique_values":
            df[column].nunique(dropna=True),

        "duplicate_values":
            len(df)
            - df[column].nunique(dropna=True)

    })


profile_df = pd.DataFrame(profile)


print("\n========== COLUMN PROFILE ==========\n")

print(profile_df)


# --------------------------------------------------
# 13. Save Profile
# --------------------------------------------------

profile_df.to_csv(
    "dataset_profile.csv",
    index=False
)

print(
    "\nProfiling completed."
)

print(
    "Profile saved as dataset_profile.csv"
)
```
